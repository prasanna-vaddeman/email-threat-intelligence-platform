# Problem Definition

## Business Problem

Organizations receive thousands of emails daily.

Some emails contain:
- spam
- phishing attempts
- malicious links
- credential theft attempts
- social engineering attacks

Manual review is slow and expensive.

We need an automated ML-powered system that can:
- classify spam emails
- identify suspicious patterns
- generate threat intelligence insights
- support monitoring and observability

## Machine Learning Problem

This is a supervised binary classification problem.

Input:
- raw email text

Output:
- HAM (safe)
- SPAM (malicious/suspicious)

Target Variable:
- spam label

Model Goal:
- accurately classify suspicious emails
- minimize false negatives

## Inputs

- email subject
- email body
- URLs
- HTML content
- suspicious keywords
- uppercase ratio
- punctuation patterns

## Outputs

- spam probability
- threat level
- threat indicators
- risk score

## Constraints

- low inference latency
- scalable API design
- real-time prediction capability
- explainability for security analysts
- monitoring support
- deployment compatibility

## Success Metrics

Primary Metrics:
- Recall
- F1-score

Secondary Metrics:
- Precision
- Accuracy

Business Metrics:
- reduced malicious email exposure
- reduced analyst workload
- faster email triage

## Baseline Model

Initial baseline:
- TF-IDF vectorization
- Logistic Regression

Why baseline first:
- simple
- interpretable
- fast
- strong NLP baseline

## Long-Term Engineering Goals

- FastAPI backend
- Streamlit frontend
- monitoring with WhyLogs
- cloud deployment
- observability
- testing pipeline
- production architecture
